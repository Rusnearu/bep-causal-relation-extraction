# R-BERT: Causal Relation Extraction (3-class)

Trains and evaluates R-BERT on CausalNewsCorpus for 3-class causal relation extraction.
Runs locally on your GPU.

## 1. Setup

```python
import os
import torch

%cd ..

# Confirm the GPU is visible before training.
print(f"CUDA available : {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"GPU            : {torch.cuda.get_device_name(0)}")
    print(f"VRAM           : {torch.cuda.get_device_properties(0).total_memory / 1e9:.1f} GB")
```

## 2. Clear tokenisation cache

```python
# R-BERT saves tokenised inputs so they are not recomputed every run.
import glob

cache_files = glob.glob("data/CausalNewsCorpus/r-bert/cached_*")
for f in cache_files:
    os.remove(f)
print(f"Cleared {len(cache_files)} cache file(s).")
```
